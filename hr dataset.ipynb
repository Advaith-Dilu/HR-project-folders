```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

#### Data loading

```python
df = pd.read_csv(r"C:\Users\Advaith's PC\OneDrive\Desktop\DATA_ANALYSIS\WA_Fn-UseC_-HR-Employee-Attrition.csv")

print(df.head())
```

#### Check Dataset Shape

```python
df.shape
```

#### View Dataset Information

```python
df.info()
```

#### Check Missing Values

```python
print(df.isnull().sum())
```

#### Description of dataset

```python
df.describe()
```

#### find Duplicates

```python
print(df.duplicated().sum())
```

#### Find Constant Columns

```python
constant_cols = [col for col in df.columns if df[col].nunique() == 1]

print(constant_cols)
```

#### Remove Constant Columns

```python
df.drop(columns=constant_cols, inplace=True)

print(df.shape)
```

#### Attrition to numeric

```python
df['Attrition'] = df['Attrition'].map({
    'Yes': 1,
    'No': 0
})

print(df['Attrition'].head())
```

#### Visualizations

#### Attrition Distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Attrition', data=df)

plt.title('Attrition Distribution')
plt.show()
```

#### Age Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20, kde=True)

plt.title('Age Distribution')
plt.show()
```

#### Gender vs Attrition

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', hue='Attrition', data=df)

plt.title('Gender vs Attrition')
plt.show()
```

#### Department vs Attrition

```python
plt.figure(figsize=(8,5))
sns.countplot(y='Department', hue='Attrition', data=df)

plt.title('Department vs Attrition')
plt.show()
```

#### Job Role vs Attrition

```python
plt.figure(figsize=(10,6))
sns.countplot(y='JobRole', hue='Attrition', data=df)

plt.title('Job Role vs Attrition')
plt.show()
```

#### Monthly Income vs Attrition

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df)

plt.title('Monthly Income vs Attrition')
plt.show()
```

#### Overtime vs Attrition

```python
plt.figure(figsize=(6,4))
sns.countplot(x='OverTime', hue='Attrition', data=df)

plt.title('OverTime vs Attrition')
plt.show()
```

#### Marital Status vs Attrition

```python
plt.figure(figsize=(7,4))
sns.countplot(x='MaritalStatus', hue='Attrition', data=df)

plt.title('Marital Status vs Attrition')
plt.show()
```

#### Education Field vs Attrition

```python
plt.figure(figsize=(10,5))
sns.countplot(y='EducationField', hue='Attrition', data=df)

plt.title('Education Field vs Attrition')
plt.show()
```

#### Work Life Balance vs Attrition

```python
plt.figure(figsize=(6,4))
sns.countplot(x='WorkLifeBalance', hue='Attrition', data=df)

plt.title('Work Life Balance vs Attrition')
plt.show()
```

#### Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(14,10))

sns.heatmap(
    numeric_df.corr(),
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

#### Top Correlation with Attrition

```python
corr = numeric_df.corr()['Attrition'].sort_values(ascending=False)

print(corr)
```
